# Exercise 3.A Getting familiar with Matplotlib 

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Creating Basic Plots

```python
x = [1, 2, 3, 4, 5] 
y = [2, 3, 5, 7, 11]
```

```python
plt.plot(x, y)
plt.show()
```

```python
plt.plot(x, y, marker ='o')
plt.show()
```

```python
plt.title('Example Graph')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.plot(x, y, marker ='o')
plt.show()
```

## Styling Plots

```python
plt.title('Example Graph')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.plot(x, y, 'o--r')
plt.show()
```

```python
plt.title('Example Graph')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.plot(x, y, ms = 10, marker = 's', mfc = 'g')
plt.show()
```

```python
plt.title('Example Graph')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.plot(x, y, '*-.m')
plt.show()
```

```python
plt.title('Example Graph')
plt.xlabel('X axis')
plt.ylabel('Y axis')
plt.plot(x, y, 'D:c', ms = 20)
plt.show()
```

# Exercise 3.B More plotting practice 

## Advanced Plotting Techniques

```python
products = ['Product A', 'Product B', 'Product C', 'Product D'] 
sales = [23, 17, 35, 29] 
```

```python
plt.title('Sales Per Product')
plt.bar(products, sales, color = 'skyblue', width=.5)
plt.xlabel('Product Name')
plt.ylabel('Sales')
```

```python
x = np.random.rand(50) 
y = np.random.rand(50)
plt.scatter(x, y)
plt.show()
```

```python
x = np.random.rand(50) 
y = np.random.rand(50)
colors1 = np.array(x) 
plt.scatter(x, y, s = np.random.rand(50) * 100, c = colors1 ,cmap='plasma')
plt.show()
```

## Subplots

```python
sine_x = np.linspace(0, 10, 100) 
sine_y = np.sin(x) 
```

```python
hist_nums = np.random.randn(100) 
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 8)) 
axes[0].plot(sine_x, sine_y, color='blue', linestyle='-') 
axes[1].hist(hist_nums, bins=15, color='skyblue', edgecolor='black') 
plt.show() 
```
