# Exercise 1.A NumPy: Manipulating Arrays  

```python
import numpy as np
```

## Looping Through Arrays

```python
arrA = np.array([[11, 12, 13],[14, 15, 16]])
print(arrA)
```

```python
for x in arrA:
    print(x * 2)
```

```python
for x in arrA:
    for y in x:
        print(y * 2)
```

The first loop prints the results as they appeared in their respective lists
and the other loop prints each result on a new line.

```python
for x in arrA:
    for y in x:
        loop_result = x * 2
        print(loop_result)
# in this loop x represents the row index and not the specific element
# so the loop returns the double of the list entirely each time
```

## Joining and Stacking Arrays

```python
joined_array_1 = np.concatenate((arrA, arrA))
print(joined_array_1)
```

```python
joined_array_2 = np.concatenate((arrA, arrA),axis = 1)
print(joined_array_2)
```

```python
stacked_array_1 = np.concatenate((joined_array_1, joined_array_1))
print(stacked_array_1)
```

```python
# stacked_array_2 = np.concatenate((joined_array_1, joined_array_2))
# print(stacked_array_1)
# ValueError: all the input array dimensions except for the concatenation axis must match exactly, 
# but along dimension 1, the array at index 0 has size 3 and the array at index 1 has size 6
```

```python
another_joined_array = np.concatenate((joined_array_1, arrA))
print(another_joined_array)
```

```python
# another_2 = np.concatenate((joined_array_1, arrA), axis = 1)
# print(another_2)
# ValueError: all the input array dimensions except for the concatenation axis must match exactly, 
# but along dimension 0, the array at index 0 has size 4 and the array at index 1 has size 2
```

```python
arr1 = np.hstack((arrA, arrA))
# stacks the arrays horizontally by column axis

arr2 = np.vstack((arrA, arrA))
# stacks the arrays vertically along row axis

arr3 = np.dstack((arrA, arrA))
# stacks the arrays by layer (height/depth)

print(arr1,'\n')
print(arr2,'\n')
print(arr3,'\n')
```

## Splitting Arrays

```python
splitsky = np.array_split(joined_array_1, 2)
print(splitsky)
# Outputs 2 arrays, each 2 dimensional
```

```python
spliterino = np.array_split(joined_array_1, 3)
print(spliterino)
```

## Searching and Replacing Array Values

```python
result_1 = np.where(arrA % 2 == 0, arrA,'odd')
print(result_1)
```

## Sorting and Filtering Arrays

```python
arrB = np.array([[7, 6], [13, 5], [7,3]])
arrB_sorted = np.sort(arrB)
print(arrB_sorted)
```

```python
print(arrB)
z = arrB < 7
arrC = arrB[z]
print(arrC)
```

```python
y = arrB >= 7
arrD = arrB[y]
print(arrD)
```
