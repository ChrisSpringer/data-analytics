# Exercise 1.B NumPy Random 

```python
import numpy as np
```

 ## Generating Random Numbers

```python
r_1 = np.random.randint(1, 20, size=(2))
print(r_1)
```

```python
r_2 = np.random.randint(1, 20, size=(1,2))
print(r_2)
```

```python
r_3 = np.random.rand(2,1)
print(r_3)
```

```python
r_4 = np.random.randint(20, 100, size=(4, 5))
print(r_4)
```

```python
r_5 = np.random.rand(4, 5)
print(r_5)
```

## Generating Random Choices from an Array

```python
students = np.array(['Ayo', 'Alvie', 'Ardita', 'Denae', 'Arionsae', 'Arber',
                   'Ajani', 'Marilyn', 'Richard', 'Jada', 'Idasha', 'Brandon'])
call_on = np.random.choice(students)
print(call_on)
```

```python
breakout_groups = np.random.choice(students, size = (3, 4))
print(breakout_groups)
# This is not an effective way to create breakout rooms since some names are repeated in the lists.
```

```python
call_on_weighted = np.random.choice(
    students, 
    p = [0, 0.04, 0.1, 0.01, 0.02, 0.11,
         0.09, 0.04, 0.3, 0.16, 0.05, 0.08],
    size = (5))
# the p values chosen were all random but still chosen so that they add up to 1
print(call_on_weighted)
```

## Generating a Normal Distribution

```python
bell_curve = np.random.normal(loc = 10, scale = 5, size=(4,5))
print(bell_curve)
```
