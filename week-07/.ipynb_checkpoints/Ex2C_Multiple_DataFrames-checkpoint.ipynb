# Exercise 2.C Working with multiple DataFrames 

```python
import numpy as np
import pandas as pd
```

## Basic Aggregations

```python
data = {
    'Product': ['A', 'B', 'C', 'A', 'B', 'C', 'A', 'B', 'C'], 
    'Region': ['North', 'South', 'East', 'West', 'North', 'South', 'East', 
               'West', 'North'], 
    'Units_Sold': [10, 15, 20, 5, 10, 15, 10, 5, 15], 
    'Price': [100, 200, 150, 100, 200, 150, 100, 200, 150]
}
df_sales = pd.DataFrame(data)
df_sales
```

```python
df_sales['Revenue'] = df_sales['Units_Sold'] * df_sales['Price']
df_sales
```

```python
print(f'Total Units Sold: {df_sales['Units_Sold'].sum()} units')
print(f'Total Revenue: ${df_sales['Revenue'].sum():,.2f}')
```

```python
print(f'Average Units Sold: {df_sales['Units_Sold'].mean():.0f} units')
print(f'Average Revenue: ${df_sales['Revenue'].mean():,.2f}')
```

 ## Grouping Data

```python
product_sales = df_sales.groupby('Product')[['Units_Sold', 'Revenue']].sum()
product_sales
```

```python
product_sales2 = (
    df_sales
    .groupby('Product')
    .agg({
        'Units_Sold': 'sum',
        'Revenue': 'sum'
    })
)

product_sales2
```

```python
region_sales = df_sales.groupby(['Region'])[['Units_Sold', 'Revenue']].mean()
region_sales
```

```python
region_sales2 = (
    df_sales
    .groupby('Region')
    .agg({
        'Units_Sold': 'mean',
        'Revenue': 'mean'
    })
)

region_sales2
```

```python
prod_reg_sales = (
    df_sales
    .groupby(['Product', 'Region'])
    .agg({
        'Units_Sold': 'sum',
        'Revenue': 'sum'
    })
)

prod_reg_sales
```

```python
prod_reg_sales2 = df_sales.groupby(['Product','Region'])[['Units_Sold', 'Revenue']].sum()
prod_reg_sales2
```

## Sorting and Ranking Data

```python
df_sales.sort_values('Units_Sold')
```

```python
df_sales.sort_values('Revenue', ascending = False)
```

```python
df_sales['Revenue_Rank'] = df_sales['Revenue'].rank(ascending = False)
df_sales
```

## Merging DataFrames

```python
product_data = {
    'Product': ['A', 'B', 'C'],
    'Category': ['Electronics', 'Furniture', 'Appliances']
} 
df_products = pd.DataFrame(product_data) 
df_products
```

```python
df_merged = pd.merge(df_sales, df_products, on=['Product'])
```

```python
df_merged
```

```python
df_merged.groupby(['Category'])[['Revenue']].sum()
```
