## Filtering a DataFrame

```python
print(df[(df['Age'] > 18)])
```

```python
filtered = df[(df['Score'] < 100) | (df['Score'] > 50)]
#filtered = df[(df['Score'] > 50)]
print(filtered)
```

 ## Modifying a DataFrame

```python
df_cleaned = df
for x in df_cleaned.index: 
    if df_cleaned.loc[x, 'Score'] < 0: 
        df_cleaned.drop(x, inplace= True) 
df_cleaned
```

```python
df_cleaned = df_cleaned.drop(columns=['City'])
df_cleaned
```

```python
df_cleaned['Age'] = df_cleaned.Age.fillna(df_cleaned.Age.mean())
df_cleaned
```

```python
df_cleaned = df_cleaned.replace({'Score':0}, df_cleaned.Score.mean())
df_cleaned
```

```python
df_cleaned = df_cleaned.dropna()
df_cleaned
```

```python
df_cleaned = df_cleaned.set_index('ID')
df_cleaned
```
