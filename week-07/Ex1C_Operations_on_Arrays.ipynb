## Sample Arrays

 ## Permutations

```python
print(r_4)
```

```python
np.random.shuffle(r_4)
print(r_4)
```

The results were shuffled by row

```python
new_shuffle = np.random.permutation(r_4)
print(new_shuffle)
print(r_4)
```

I would use permutation when I want a shuffled copy of the array without altering the order of the original array.
I would uuse shuffle if I need a random ordering of the contents.

```python
print(r_1)
print(r_2)
print(np.shape(r_1))
print(np.shape(r_2))
print(np.add(r_1,r_2))
print(np.shape(np.add(r_1,r_2)))
```

```python
print(r_1)
print(r_3)
print(np.shape(r_1))
print(np.shape(r_3))
print(np.add(r_1,r_3))
print(np.shape(np.add(r_1,r_3)))
```

```python
print(r_2)
print(r_4)
print(np.shape(r_2))
print(np.shape(r_4))
# print(np.add(r_2,r_4))
# ValueError: operands could not be broadcast together with shapes (1,2) (4,5) 
# They cannot be added together due to their shapes
```

```python
print(np.add(r_4, np.array([1])))
```

```python
print(np.add(r_4, np.array([[1]])))
```

```python
print(np.add(r_4, np.array([1,1,1,1,1])))
```

```python
# print(np.add(r_4, np.array([[1],[1]])))
# ValueError: operands could not be broadcast together with shapes (4,5) (2,1)
```

Broadcasting is able to combine elements in 2 arrays without the need of a for loop.
Dimensions align from the right and are compatible if they are equal, or one of them is 1.
It fails when any dimension pair is neither equal nor 1, (1, 5 could work, but 4,2 does not in the above example)
Numpy stretches dimensions, or repeats values to match shapes.
The array with the larger shape determine the shape of the resulting array.

## Grocery Problem

```python
bag1 = [2, 3, 4, 0] # two vegetables, three fruits, four boxes of snacks, and no beverages
bag2 = [4, 0, 1, 2] # no fruits, four vegetables, two beverages, and one box of snacks print
print(bag1,'\n',bag2)
```

```python
total_items1 = np.sum(bag1)
total_items2 = np.sum(bag2)
print(f'The first bag contains {total_items1} items, and the second bag contains {total_items2} items.')
```

```python
total = np.add(bag1, bag2)
print(f'There are {total[0]} vegetables, {total[1]} fruits, {total[2]} boxes \
of snacks, and {total[3]} beverages in the two grocery bags.')
```

## Teamwork

```python
ali = [5,4,6,4,2]
```

```python
ben = [0,5,5,3,4]
```

```python
carla = [3,6,0,3,4]
```

```python
all_knocks = [ali,ben,carla]
print(all_knocks)
```

```python
print(f'\
For the week, Ali knocked on {np.sum(all_knocks[0])} doors, \
Ben knocked on {np.sum(all_knocks[1])} doors, \
and Carla knocked on {np.sum(all_knocks[2])} doors. \
In total the group knocked on {np.sum(all_knocks)} doors.')
```

```python
total_votes = np.sum(all_knocks, axis = 0)
total_votes = np.where(total_votes > 0 , total_votes // 2, total_votes)
print(f'Estimated Votes:\
Monday:      {total_votes[0]} votes \
Tuesday:     {total_votes[1]} votes \
Wednesday:   {total_votes[2]} votes \
Thursday:    {total_votes[3]} votes \
Friday:      {total_votes[4]} votes \
Total votes: {np.sum(total_votes)} votes')
```
