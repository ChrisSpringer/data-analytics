# Exercise 2.C Creating and Calling Functions

## Lab 3. - tax functions

```python
def get_soc_sec_tax(amount):
    tax_rate = 0.062
    return amount * tax_rate
```

```python
def get_medicare_tax(amount):
    tax_rate = 0.0145
    return amount * tax_rate
```

```python
def get_federal_tax(amount, code):
    if code == 0:
        tax_rate = 0.23
    elif code == 1:
        tax_rate = 0.21
    elif code == 2:
        tax_rate = 0.195
    elif code == 3:
        tax_rate = 0.185
    elif code == 4:
        tax_rate = 0.18
    else:
        tax_rate = 0.18 - ((code - 4) * .005)
    return amount * tax_rate
```

```python
print(f'Person 1:  gross pay $750 Social Security Tax ${get_soc_sec_tax(750):,.2f}')
print(f'Person 2:  gross pay $1550 Social Security Tax ${get_soc_sec_tax(1550):,.2f}')
print(f'Person 3:  gross pay $1100 Social Security Tax ${get_soc_sec_tax(1100):,.2f}')
```

```python
print(f'Person 1:  gross pay $750 Medicare Tax ${get_medicare_tax(750):,.2f}')
print(f'Person 2:  gross pay $1550 Medicare Tax ${get_medicare_tax(1550):,.2f}')
print(f'Person 3:  gross pay $1100 Medicare Tax ${get_medicare_tax(1100):,.2f}')
```

```python
print(f'Person 1:  gross pay $750  withholding code 0 Federal Tax ${get_federal_tax(750, 0):,.2f}')
print(f'Person 2:  gross pay $1550  withholding code 2 Federal Tax ${get_federal_tax(1550, 2):,.2f}')
print(f'Person 3:  gross pay $1100  withholding code 5 Federal Tax ${get_federal_tax(1100, 5):,.2f}')
```
