# Exercise 2.C Creating and Calling Functions

## Lab 1. - working with defining custom functions

```python
def favorite_things(name, movie, music):
    return f'Your name: {name}\nYour favorite movie: {movie}\n\
Your favorite musician/band: {music}'
print(favorite_things('Chris', 'Shrek', 'Daft Punk'))
```

```python
def why_im_here(purpose):
    return f'Your reason for joining YearUp United Data Analyst program is: {purpose}.'
print(why_im_here("To acquire an internship and help gain \
experience to help transiton into the workforce"))
```

```python
def favorite_place(place, why):
    print(f'One of my favorite places is {place} because {why}.')
```

```python
favorite_place('my home', 'it is quiet, peaceful, and I can relax')
```

```python
favorite_place('the mall', 'I like to shop and buy new things and try new foods')
```
