# Exercise 2.A Using built-in functions

## Lab 2. Parsing strings - part codes

```python
def get_supplier_code(code):
    supplier_code = code[:code.find(":")]
    return supplier_code
```

```python
def get_product_num(code):
    product_num = code[code.find(":") + 1:code.find("-")]
    return product_num
```

```python
def get_size(code):
    size = code[code.find("-") + 1:]
    return size
```

```python
print(f'For {'ACME:123-L'}, the supplier is {get_supplier_code('ACME:123-L')}. \
Part number {get_product_num('ACME:123-L')} is available in size {get_size('ACME:123-L')}.')
```

```python
print(f'For {'DI:12345-M'}, the supplier is {get_supplier_code('DI:12345-M')}. \
Part number {get_product_num('DI:12345-M')} is available in size {get_size('DI:12345-M')}.')
```

```python
print(f'For {'ACE:1-12'}, the supplier is {get_supplier_code('ACE:1-12')}. \
Part number {get_product_num('ACE:1-12')} is available in size {get_size('ACE:1-12')}.')
```
