## Lab 1. Working with names - parsing strings

```python
test_name = 'John Doe'
```

```python
def parse_and_display(name):
    for i in range(len(name)):
        if name[i] == " ":
            first = name[:i]
            last = name [i+1:]
            print(f"Full name:  {name}")
            print(f"First name: {first}")
            print(f"Last name:  {last}")
```

```python
parse_and_display(test_name)
```

```python
test_list = ['Grace Flores', 'JB Oinonen', 'Ken Penn']
```

```python
parse_and_display(test_list[0])
```

```python
parse_and_display(test_list[1])
```

```python
parse_and_display(test_list[2])
```

```python
parse_and_display('Christopher Springer')
```

```python
def parse_and_display2(name):
    contains_space = False
    for i in range(len(name)):
        if name[i] == " ":
            contains_space = True
            first = name[:name.find(" ")]
            middle = name[name.find(" ") + 1: name.rfind(" ")]
            last = name[name.rfind(" ") + 1:]
            print(f"Full name:  {name}")
            print(f"First name: {first}")
            if middle != '':
                print(f"Middle name: {middle}")
            print(f"Last name: {last}")
            break
    if contains_space == False:
        print(f"Full name:  {name}")
```

```python
test_list2 = ['Lorde', 'Billie Eilish', 'Megan Thee Stallion']
```

```python
parse_and_display2(test_list2[0])
```

```python
parse_and_display2(test_list2[1])
```

```python
parse_and_display2(test_list2[2])
```
