# Exercise 2.B Using Library Functions
## Lab 2. Working with datetime functions

```python
import datetime
```

# Today

```python
today = datetime.datetime.now()
```

```python
print(f'Today is {today.strftime('%A')}, {today.strftime('%B')} \
{today.strftime('%d')}, {today.strftime('%Y')}')
```

# My birthday

```python
birthday = datetime.date(1997, 1, 5)
```

```python
print(f'My birthday is {birthday.strftime('%x')}')
```

# Next 90 Days

```python
ninety_d = today + datetime.timedelta(days = 90)
```

```python
print(f'90 days from today is {ninety_d.strftime('%B')} \
{ninety_d.strftime('%d')}, {ninety_d.strftime('%Y')}')
```

# Time for Dinner

```python
dinner_time = datetime.time(18, 30)
```

```python
print(f"Let's meet for dinner at {dinner_time.strftime('%X')}")
```
